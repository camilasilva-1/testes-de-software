```python
%%writefile sistema.py

def obter_dados_turma():
  nomes = ["Ana", "Bruno", "Carla", "Diego", "Elisa"]
  notas_1av = [7.5, 6.0,  8.5,  4.0,  9.0]
  notas_2av = [8.0, 5.5,  7.0,  3.5,  9.5]
  return nomes,notas_1av,notas_2av
```

```python
%%writefile test_sistema.py

from sistema import obter_dados_turma

def test_obter_dados_turma_retorna_tre_lista():
 nomes,notas_1av,notas_2av = obter_dados_turma()

 assert isinstance(nomes, list), "nomes deve ser uma lista"
 assert isinstance(notas_1av, list), "notas_1av deve ser uma lista"
 assert isinstance(notas_2av, list), "notas_2av deve ser uma lista"

def test_obter_dados_turma_tem_cinco_alunos():
  nomes, notas_1av, notas_2av = obter_dados_turma()
  assert len(nomes) == 5, "verificar se a lista de nomes tem 5 alunos matriculados"

def test_obter_dados_turma_lista_do_mesmo_tamanho():
  nomes, notas_1av, notas_2av = obter_dados_turma()
  assert len(nomes) == len(notas_1av)
  assert len(nomes) == len(notas_2av)



```
